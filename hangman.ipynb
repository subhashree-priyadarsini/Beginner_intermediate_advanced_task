# Hangman - Word Guessing Game

## How to play
1. Run the code cell below to start a game. A secret word is chosen at random, and its hint is shown at the beginning.
2. When prompted with **Enter a letter**, type one letter and press **Enter**.
3. Correct guesses reveal that letter in the word. Each incorrect guess adds a part to the hangman drawing and uses one attempt.
4. You have 6 incorrect guesses. Guess every letter before they run out to win; otherwise, the game shows the answer.
5. If you enter a letter you already guessed, the game says **Already guessed!** and does not use another attempt.
6. Run the code cell again to start a new game.

```python
import random

words = {
    "python": "Programming Language",
    "apple": "A fruit",
    "tiger": "Wild animal",
    "india": "A country",
    "computer": "Electronic device"
}

stages = [
"""
 -----
 |   |
 |
 |
 |
 |
------
""",
"""
 -----
 |   |
 |   O
 |
 |
 |
------
""",
"""
 -----
 |   |
 |   O
 |   |
 |
 |
------
""",
"""
 -----
 |   |
 |   O
 |  /|
 |
 |
------
""",
"""
 -----
 |   |
 |   O
 |  /|\\
 |
 |
------
""",
"""
 -----
 |   |
 |   O
 |  /|\\
 |  / \\
 |
------
""",
"""
 -----
 |   |
 |   O
 |  /|\\
 |  / \\
 |
------
"""
]

word = random.choice(list(words.keys()))
hint = words[word]

guessed_letters = []
wrong_guesses = 0
max_attempts = 6

print("=== HANGMAN GAME ===")
print("Hint:", hint)

while wrong_guesses < max_attempts:

    display_word = ""

    for letter in word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "

    print("\nWord:", display_word)

    if "_" not in display_word:
        print("🎉 Congratulations! You Won!")
        break

    guess = input("Enter a letter: ").lower()

    if guess in guessed_letters:
        print("Already guessed!")
        continue

    guessed_letters.append(guess)

    if guess not in word:
        wrong_guesses += 1

        print("Wrong Guess!")
        print(stages[wrong_guesses])

        print("Remaining Attempts:",
              max_attempts - wrong_guesses)

if wrong_guesses == max_attempts:
    print("\n😢 Game Over")
    print("Correct Word:", word)
```
